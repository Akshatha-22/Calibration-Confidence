```python
import os

tuning_base_path = 'results/tuning/'

if not os.path.exists(tuning_base_path):
    print(f"No tuning results found at {tuning_base_path}.")
else:
    model_directories = [d for d in os.listdir(tuning_base_path) if os.path.isdir(os.path.join(tuning_base_path, d))]

    if not model_directories:
        print(f"No model tuning directories found under {tuning_base_path}.")
    else:
        print("Available tuning results:")
        for model_name in sorted(model_directories):
            model_path = os.path.join(tuning_base_path, model_name)
            tuning_runs = [d for d in os.listdir(model_path) if os.path.isdir(os.path.join(model_path, d))]
            if tuning_runs:
                print(f"  Model: {model_name}")
                for run in sorted(tuning_runs, reverse=True):
                    print(f"    - {run}")
            else:
                print(f"  Model: {model_name} (No tuning runs found)")
```

```python
import os

# List the directories under results/tuning/deep/
deep_tuning_dirs = [d for d in os.listdir('results/tuning/deep/') if os.path.isdir(os.path.join('results/tuning/deep/', d))]

if deep_tuning_dirs:
    print(f"Found Deep MLP tuning directories: {deep_tuning_dirs}")
    # Assuming the most recent one is the one to inspect
    latest_deep_run = sorted(deep_tuning_dirs, reverse=True)[0]
    print(f"Using the latest Deep MLP tuning run: {latest_deep_run}")
else:
    print("No Deep MLP tuning directories found. You might need to run Deep MLP tuning first.")
```
